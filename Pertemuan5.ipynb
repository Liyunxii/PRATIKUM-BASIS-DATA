**5.3 INSERT TABLE**

Perintah *INSERT INTO *digunakan untuk menambah data *(record)* baru ke dalam tabel.

*   Sintaks Dasar: *Insert Into Nama_Table Values (nilai1, nilai2, ...);*  (Implied).

**5.4 QUERY SEDERHANA (SELECT)**

Perintah SELECT merupakan perintah utama DML yang digunakan untuk:

*   Menampilkan data secara keseluruhan yang terdapat di dalam tabel.
*   Menampilkan data tertentu yang terdapat di dalam tabel.
*   Menampilkan dan mengurutkan data secara ascending dan descending.

**Menampilkan Data Secara Keseluruhan (5.4.1):**

*   Sintaks: Select * From Nama_Table;.

**Menampilkan Kolom Data Tertentu:**

*   Sintaks: Select Nama_Field_1, … , Nama_Field_N From Nama_Table;.

**5.5 UPDATE TABLE**

Perintah UPDATE digunakan untuk merubah data (record) yang sudah ada di dalam tabel.

**5.6 DELETE DATA**

Perintah DELETE digunakan untuk menghapus data (record) di dalam tabel.

**5.7 QUERY DENGAN KONDISI**

Query dapat ditambahkan kondisi (WHERE) untuk membatasi baris data yang ditampilkan atau dimanipulasi.

*   Bentuk Umum: Select * from Nama_Table where Kondisi;.
*   Operator Relasional: Kondisi menggunakan operator relasional (=, >, <, >=, <=, <>) untuk membandingkan nilai.
*   Contoh: select * from customer where cust_id = ‘10003’ or cust_name =’wascals; (menampilkan data customer yang memenuhi salah satu dari dua kondisi).

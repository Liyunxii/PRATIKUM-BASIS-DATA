**BAB 4: ALTER, MODIFY, DROP, RENAME (DDL)**

**4.1 IDENTITAS**

Kompetensi yang ditargetkan pada bab ini adalah agar praktikan memahami penggunaan DDL (Data Definition Language) antara lain ALTER, MODIFY, DROP, dan RENAME.

*Topik yang dibahas meliputi:*

1.   ALTER, MODIFY.
2.   DROP, RENAME.

**4.3 ALTER TABLE**

Perintah ALTER TABLE digunakan untuk merubah struktur tabel yang sudah ada. Ada empat macam perubahan yang dapat dilakukan terhadap struktur tabel:

1.   Merubah Nama Field/Kolom

*   Perubahan yang terjadi hanya pada nama field/kolom.
*   Digunakan perintah Change.
*   **Sintaks Perintah:** Alter Table Nama_Table Change Nama_Field_Lama Nama_Field_Baru Tipe_Data (Size);

2.   Merubah Tipe Data

*   Perubahan yang terjadi hanya pada tipe data yang digunakan oleh field/kolom tertentu.
*   Digunakan perintah Modify.
*   **Sintaks Perintah:** Alter Table Nama_Table Modify Nama_Field_Lama Tipe_Data_Baru (Size);

3.   Penambahan Field

*   Digunakan untuk menambahkan field/kolom baru ke dalam tabel.
*   Digunakan perintah Add.
*   **Sintaks Perintah:** Alter Table Nama_Table Add Nama_Field_Baru Tipe_Data (Size);

4.   Penghapusan Field

*   Digunakan untuk menghapus field/kolom tertentu dari tabel.
*   Digunakan perintah Drop.
*   **Sintaks Perintah:** Alter Table Nama_Table Drop Nama_Field_Lama;

**4.4 DROP (Menghapus Table)**

*   Perintah Drop Table digunakan jika tabel yang telah dibuat sudah tidak dibutuhkan lagi dan ingin dihapus secara permanen.
*   **Sintaks Perintah:** Drop Table Nama_Table;
